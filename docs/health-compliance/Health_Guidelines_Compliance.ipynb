## Doc 1 - EMR System Implementation Guidelines (extracts)

**Governance:** public rollout needs MoH approval via EMRS Working Group; private
implementers follow MoH guidance. Quarterly self-assessments + annual external
assessment are mandatory for ALL facilities including private.

**Minimum functional scope:** patient registration with unique ID, internal +
external referrals, stock management, disease surveillance reporting, real-time
analytics and aggregate reports, save-blocking validation rules, decision support,
system administration and security.

**Design principles:** usability, scalability, availability, interoperability,
open standards, data quality, ethics, access control.

**Security:** RBAC mandatory; no credential sharing; least privilege (facility
staff see own-facility records, district/national see reports only); periodic
security assessments; official emails only for reports; test-only-in-test-env
change control with backups before production change; downtime SOP with paper
fallback and 24h back-entry.

**Privacy:** comply with Act 2019 + Protection Guidelines; data requests go to
the Director General MoH; breaches reported to PDPO + MoH immediately; government
owns all health data (private facility = trustee); patients access records free
(printing only).

**Interoperability:** comply with Exchange/Interoperability Guidelines; integrate
DHIS2/eHMIS + NHIE registries (Facility, Products, Workers, Client); private
certified EMRS must produce HMIS reports + birth/death notifications; emergency
client-level submission during outbreaks. Standards: HL7/FHIR, ICD, SNOMED CT.

**Data:** automated offsite backups, daily off-peak, monthly restore tests;
tamper-proof audit logs of ALL activity incl. access/modify/delete, regularly
reviewed; retention per Protection Guidelines Sec 5.6; HMIS push every 7th,
review 8th-15th, DHIS2 locked after 15th.

**Delivery:** internal tests + signed UAT per release; same version everywhere;
training on an isolated instance (never production); min 5-day worker training
with 70% pass mark; 5-tier support escalation (call center 0800-100-066).


## Doc 2 - Digital Health Solutions Introduction (extracts)

**Approval chain (sequential, vendor never self-presents - User Dept sponsors):**
DHSC registration (dhs@health.go.ug / dhtr.go.health.ug, 24h-5d ack) -> DHSC
presentation (DHSAC scored, report in 3 weeks) -> field visit if required ->
HIIRE TWG (pilot ONLY) -> SMC (scale/policy pilots) -> HPAC (policy/finance)
-> Top Management (procurement/scale/adoption). Live unapproved systems must
re-enter retroactively. Scale needs 6 dossiers (transition, sustainability,
capacity, specification, requirements, scale-up plans).

**DHSAC security bar:** external pen test incl. OWASP Top 10 within 12 months
with NO vulnerability >= 7.0 CVSS; custom code review; audit trails of all
access; load testing.

**Data:** declare PID access (then Act 2019 applies); declare storage location
(Uganda vs offshore + country + justification); expose documented open APIs
(HL7/FHIR, OAuth 2.0/TLS 1.2); unique patient identifier; DHIS2 HMIS push mandatory.

**Usability evidence:** user research, journey maps, signed UAT, agile iteration.
**Sustainability:** cost declaration + sustainability plan scored.


## Doc 3 - Health Data Protection Guidelines (extracts)

**Consent:** before collection; free, informed (privacy notice), specific,
affirmative, withdrawable as easily as given; fresh consent on policy change;
parental consent + proofs for children (assent/confidentiality exceptions);
nominee hierarchy for incapacitated (named > family > authorized > clinician).
Provider-to-provider treatment sharing needs NO consent (vital interests).

**Privacy notice (Appendix II):** before collection + on change + new purpose;
physical display + website + service languages; 9 content items (extent, data
nature, methods, controller identity, withdrawal, recipients, retention, rights,
complaints).

**Access control:** least-necessary default; role matrix; strong passwords +
MFA for privileged/network; session lock/timeout; failed-logon limits; remote
session encryption; endpoint + physical controls.

**Encryption:** PII encrypted at rest, in transit, on media and backups;
FIPS-validated crypto; de-identification for analysis; encrypted offsite backups
+ restore tests.

**Breach:** procedure publicly displayed; PDPO notified IMMEDIATELY; subjects
within 24h if high risk; processor to controller without undue delay.

**Retention:** minimum 5 years from creation; disposal per archives law;
processor deletes all copies within 10 business days of cessation + certifies.
Erasure with care; block/restrict where law mandates retention.

**Rights (via DPO, 5-working-day SLA; corrections decided in 7 days;
grievance ack 7 days, resolve 1 month):** access/export, rectification with
versioning (original preserved), erasure/blocking, restriction, automated-
decision opt-out, nominee/deceased gates, downstream notification.

**Governance:** PDPO registration; DPO at institutional level; DPIA before new
high-risk tech; annual independent audit; yearly mandatory staff training
(access revoked if missed); no cross-border storage without PDPO auth +
adequacy + MoH consent; sale of health data prohibited (Sec 37).


## Doc 4 - Data Protection and Privacy Act, 2019 (extracts)

Applies to anyone collecting/processing in Uganda + anyone abroad handling
Ugandan citizens' data. Enforced by PDPO under NITA. Health data = special
personal data: collection PROHIBITED except 3 gateways - legal duty, free
consent, or qualifying non-profit activity. 'Medical purposes' allows narrow
no-consent processing (treat as direct-care/emergency only).

**Duties:** prior consent (children: parental); 9-item pre-collection notice;
purpose lock (secondary use needs fresh consent or Sec 17(3) gateway);
minimisation; accuracy + correction with downstream notification; retention
only as allowed + secure destroy/de-identify; infosec with verification;
processor DPAs; immediate Authority breach notice (Authority directs subject
notice); access in 30 days; stop-processing in 14/7 days; marketing opt-out
14 days; automated-decision review 21/14/7 days; DPO designation; register
with Authority; cross-border only with adequacy or consent.

**Penalties (1 point = UGX 20,000):** unlawful obtain/disclose <= 240pts
(UGX 4.8M) and/or 10y (Sec 35); unlawful destroy/alter >= 240pts and/or 10y
(Sec 36); sale <= 245pts (UGX 4.9M) and/or 10y (Sec 37); corporate + officers
liable + up to 2% turnover (Sec 38); plus remedial orders + compensation.
